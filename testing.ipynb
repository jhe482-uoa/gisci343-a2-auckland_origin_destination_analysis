```python
import pandas as pd
import geopandas as gpd
from ipyleaflet import Map, GeoData, LayersControl, basemaps, basemap_to_tiles
```

```python
df = pd.read_csv(r"data\statsnz-2023-census-main-means-of-travel-to-work-by-statistical-area-CSV\2023-census-main-means-of-travel-to-work-by-statistical-area.csv")

df
```

```python
df = df.replace(-999,0)
```

```python
list(df["SA22023_V1_00_NAME_ASCII_usual_residence_address"].unique())
```

```python
def obtain_movement(df, mode, target):
    # Mode = Leaving from
    if mode:
        filtered = df[df["SA22023_V1_00_NAME_ASCII_usual_residence_address"] == target]
        return filtered
    # Mode = Coming to
    else:
        filtered = df[df["SA22023_V1_00_NAME_ASCII_workplace_address"] == target]
        return filtered
```

```python
obtain_movement(df, 0, "Newmarket")
```

```python
path = r"data\statsnz-statistical-area-2-2023-generalised-SHP\statistical-area-2-2023-generalised.shp"
sa2shape2023 = gpd.read_file(path)
new_tiles = basemap_to_tiles(basemaps.CartoDB.Positron)

sa2shape2023 = sa2shape2023.to_crs(epsg=4326)
center_x = sa2shape2023.geometry.centroid.x.mean()
center_y = sa2shape2023.geometry.centroid.y.mean()
map = Map(center=(center_y, center_x), zoom=5, layers=(new_tiles, ))

geo_data = GeoData(geo_dataframe=sa2shape2023, style={'color': 'white', 'fillOpacity': 0.3, 'weight': 1},
    name='NZ Regions')

map.add_layer(geo_data)
display(map)
```
